### **TASK3**: **Data Quality check**

**1.For each state calculate the total number of transactions.total transaction** **amount and total registered users by summing up the values from** **the district level data**

```python
#district level
gp=df4.groupby(["State"],as_index=False)[["Transactions","Amount (INR)","Registered Users"]].sum()
display(gp)
```

**2.compare the results with the corresponding values at the state level to** **ensure they match**

```python
#state-level data
gp1=df1.groupby(["State"],as_index=False)[["Transactions","Amount (INR)","Registered Users"]].sum()
compare=pd.merge(gp,gp1,on=["State"],suffixes=["_district_level","_state_level"])
display(compare)
```

**3.display any discrepancies found between the district-level and state-level** **data**

```python
compare["count_diff"]=compare["Transactions_district_level"]-compare["Transactions_state_level"]
display(compare["count_diff"])
compare["amount_diff"]=compare["Amount (INR)_district_level"]-compare["Amount (INR)_state_level"]
display(compare["amount_diff"])
compare["user_diff"]=compare["Registered Users_district_level"]-compare["Registered Users_state_level"]
display(compare["user_diff"])
```

```python
discrepancies = compare[
    (compare["count_diff"] != 0) |
    (compare["amount_diff"] != 0) |
    (compare["user_diff"] != 0)
]
display(discrepancies)
```

**4.Task 4:Data merging and advance analysis**

*   **4.1 Ratio of users to population by state**
*   **1.merge the state_Txn and Users dataset with the District Demographics** **dataset to calculate the ratio of registered users to the population for each** **state display the result in tabluar format**



```python
state_user=df1.groupby(["State"],as_index=False)["Registered Users"].sum()
state_population=df5.groupby(["State"],as_index=False)["Population"].sum()
merge=pd.merge(state_user,state_population,on="State",how="inner")
merge["user_to_population_ratio"]=merge["Registered Users"]/merge["Population"]
display(merge[["State","Registered Users","Population","user_to_population_ratio"]])
```

**2.Create column chart depicting the ratio of users to population by state**

```python
plt.figure(figsize=(12,6))
plt.bar(merge["State"],merge["user_to_population_ratio"],color="skyblue")
plt.xlabel("STATE",fontsize=12)
plt.title("Ratio of Registered Users to Population by State",fontsize=14,fontweight="bold")
plt.ylabel("User-to-Population Ratio",fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
```
